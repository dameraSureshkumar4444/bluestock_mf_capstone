```python
import numpy as np
import pandas as pd
import os
df = pd.read_csv('../data/raw/02_nav_history.csv')

# datetime
df["date"] = pd.to_datetime(df["date"])

# Sort
df = df.sort_values(["amfi_code", "date"])

# Remove duplicates
df = df.drop_duplicates()

# NAV must be positive
df = df[df["nav"] > 0]

# Forward fill NAV within each fund
df["nav"] = df.groupby("amfi_code")["nav"].ffill()

os.makedirs("data/processed", exist_ok=True)
df.to_csv(r"c:\Users\Damerasureshkumar\OneDrive\Desktop\bluestock_mf_capstone\data\processed\02_nav_history_cleaned.csv", index=False)

print(df.shape)
```
