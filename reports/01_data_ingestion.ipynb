```python
import pandas as pd

fund_master = pd.read_csv("../data/raw/01_fund_master.csv")
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")

print("Fund Master Codes:", fund_master["amfi_code"].nunique())
print("NAV History Codes:", nav_history["amfi_code"].nunique())

master_codes = set(fund_master["amfi_code"])
nav_codes = set(nav_history["amfi_code"])

missing_codes = master_codes - nav_codes

if len(missing_codes) == 0:
    print("\nSUCCESS: All AMFI codes from fund_master exist in nav_history")
else:
    print("\nMissing Codes:")
    print(missing_codes)

```

```python
merged = fund_master.merge(
    nav_history[['amfi_code']].drop_duplicates(),
    on="amfi_code",
    how="left",
    indicator=True
)
merged
```

```python
from pathlib import Path

Path("data").mkdir(parents=True, exist_ok=True)
out_path = Path("..") / "data" / "processed" / "merged.csv"
out_path.parent.mkdir(parents=True, exist_ok=True)
merged.to_csv(out_path, index=False)
```
